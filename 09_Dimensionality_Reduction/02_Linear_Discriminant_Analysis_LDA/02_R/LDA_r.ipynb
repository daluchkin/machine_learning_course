# Linear Discriminant Analysis (LDA)

```r
suppressMessages(library(tidyverse))
suppressMessages(library(ggplot2))
suppressMessages(library(caret))
suppressMessages(library(e1071))
suppressMessages(library(MASS))

options(repr.plot.width=12, repr.plot.height=8)
```

## Dataset

```r
data <- read_csv('../00_data/Wine.csv', show_col_types = FALSE)

head(data)
```

```r
glimpse(data)
```

## Splitting the data into training and test set

```r
set.seed(42)

train_data <- data |>
                slice_sample(prop = 0.8)
test_data <- data |>
            anti_join(train_data, by = colnames(data))
```

```r
dim(train_data)
```

```r
dim(test_data)
```

## Feature Scaling

```r
train_data[-ncol(train_data)] <- scale(train_data[-ncol(train_data)])
test_data[-ncol(test_data)] <- scale(test_data[-ncol(test_data)])
```

```r
head(train_data)
```

## Applying LDA

```r
lda <- lda(Customer_Segment ~ ., data = train_data)
train_data <- as.data.frame(predict(lda, train_data))
train_data <- train_data[c(5, 6, 1)]
test_data <- as.data.frame(predict(lda, test_data))
test_data <- test_data[c(5, 6, 1)]

print(head(train_data))
print(head(test_data))
```

## Training the model on the training set

```r
fit <- svm(class ~ x.LD1 + x.LD2, data = train_data, type = "C-classification", kernel = "linear")
summary(fit)
```

```r
# Evaluation
y_pred <- predict(fit, test_data)
y_pred
```

## Model Performance

### Confusion Matrix

```r
table(test_data$class, y_pred)
```

### Accuracy

```r
accuracy <- mean(test_data$class == y_pred)
print(paste('Accuracy:', round(accuracy*100, 2), "%"))
```

## Viz the results

```r
svm_pred_boundary <- function(df_train, df_test, name){
    X1 <- seq(min(df_test$x.LD1) - 1, max(df_test$x.LD1) + 1, length.out = 200)
    X2 <- seq(min(df_test$x.LD2) - 1, max(df_test$x.LD2) + 1, length.out = 200)
    grid <- expand.grid(X1, X2)
    colnames(grid) <- c('x.LD1', 'x.LD2')
    grid$Prediction <- predict(fit, newdata = grid)
    
    print(ggplot() +
        geom_tile(data=grid, aes(x = x.LD1, y = x.LD2, fill = as.factor(Prediction)), alpha=0.4) +
        geom_point(df_test, mapping=aes(x = x.LD1, y = x.LD2, colour = class), size=2) +
        scale_fill_manual(values = c("red", "blue", "orange"), name = "Prediction") +
        scale_color_manual(values = c("red", "blue", "orange"), name = "class") +
        labs(title = paste0("SVM (", name,")"), x = "LD1", y = "LD2") +
        theme_minimal())
}
```

```r
svm_pred_boundary(train_data, train_data, 'Training set')
```

```r
svm_pred_boundary(test_data, test_data, 'Test set')
```
